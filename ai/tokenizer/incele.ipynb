# Tokenizer incelemesi (M5)
Türkçe kelimeler nasıl bölünüyor, kart etiketleri tek token mı, bölünüp birleştirilen kart hâlâ okunuyor mu?

Önce eğit: `python -m ai.tokenizer.egit_tokenizer --girdi …` (bkz. `ai/tokenizer/README.md`).

```python
import os, sys
if os.path.basename(os.getcwd()) == 'tokenizer': os.chdir('../..')   # AI/ klasöründen çalış
sys.path.insert(0, os.getcwd())
from tokenizers import Tokenizer
from ai.tokenizer import incele
tok = Tokenizer.from_file(str(incele.CIKTI))
print('kelime dağarcığı:', tok.get_vocab_size())
```

## 20 örnek cümle

```python
for m in incele.ORNEKLER:
    t = incele.bolunus(tok, m)
    print(f'{len(t):3d} · {m}\n      ' + ' | '.join(t))
```

## Kart etiketleri ve gidiş-dönüş

```python
print('tek token olmayan etiketler:', incele.etiket_kontrol(tok) or 'yok ✓')
geri = tok.decode(tok.encode(incele.ORNEK_KART).ids, skip_special_tokens=False)
print(geri)
print('okunuyor:', incele.metinden_karta(geri)[0] == incele.metinden_karta(incele.ORNEK_KART)[0])
```
